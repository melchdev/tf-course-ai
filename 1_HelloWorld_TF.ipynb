Considere el siguiente conjunto de reglas:

```
def hw_function(x):

    y = (2 * x) - 1
    return y
```

Usando el paradigma de aprendizaje de máquina, intentaremos "adivinar las reglas" dadas una entradas y unas salidas (Esperadas).

Importamos Tensorflow y Numpy.

OJO: Recomendado inicializar un ambiente virtual (O pueden aparecer problemas de dependencias).

````python
import numpy as np
import tensorflow as tf
````

Definimos los siguientes arreglos

````python
xs = np.array( [-1.0, 0.0, 1.0, 2.0, 3.0, 4.0], dtype = float )
ys = np.array( [-3.0, -1.0, 1.0, 3.0, 5.0, 7.0], dtype = float )
````

Ahora, procedemos a crear y compilar una red neuronal. Una simple, que sólo se compone de 1 capa y una neurona. Usar comando [Sequential](https://keras.io/api/models/sequential/) que será un conjunto de capas ( [Layers](https://keras.io/api/layers/) ), y a su vez, son el bloque básico en una red neuronal en un modelo. Se puede usar una capa "Densa" ( [Dense](https://keras.io/api/layers/core_layers/dense/) ).

También, es buena práctica indicar el tamaño de la entrada al modelo. Usar el comando [tf.keras.Input()](https://www.tensorflow.org/api_docs/python/tf/keras/Input) con el parámetro `shape`, esto no es una capa adicional, debe ir dentro de la capa.

````python
model = tf.keras.Sequential([

    # Tamaño de la entrada
    tf.keras.Input( shape=(1,) ),

    # Capa densa
    tf.keras.layers.Dense( units = 1 ) # Units, dimensión del tamaño de salida
    # Tiene muchos parámetros, como activation, use_bias...
    
])
````

Ahora, hay que definir dos funciones: Función de pérdida ( [Loss](https://keras.io/api/losses/) ) y un optimizador ( [Optimizer](https://keras.io/api/optimizers/) ). Hay varios para escoger. Se requiere un análisis cuidadoso del problema, y una compresión profunda sobre la estadística y matemática dentro. 

Sin embargo, siempre se puede experimentar y jugar un poco con ellos!

````python
model.compile( optimizer = 'sgd', loss = 'mean_squared_error' )
# Se escoge un Stochastic Gradient Descent (SGD) y un error cuadrático medio.
````

Finalmente, queda entrenar la red neuronal. Usar la famosa función `model.fit(x,y)`.
Se usa el optimizador para ir mejorando los parámetros y así, el rendimiento del modelo. Se hará por el número de `epochs` específicado.

````python
model.fit(xs, ys, epochs=500)
````

Al predecir un valor, no necesariamente obtenemos el valor exacto. Esto es que, en ML y DL se trabaja con probabilidades, no con modelos deterministas.

````python
print( f"Predicción: {model.predict(np.array([20.0]), verbose = 0).item():.2f}" )
````
